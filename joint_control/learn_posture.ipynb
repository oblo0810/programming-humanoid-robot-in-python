use machine learning to recognize robot's posture (following the example in [scikit-learn-intro.ipynb](./scikit-learn-intro.ipynb) )

## 1. Data collection

We have colleceted data before, you need to add new data if you want to add new posture.

* the dateset are in *robot_pose_data* folder
* each file contains the data belongs to this posture, e.g. the data in *Back* file are collected when robot was in "Back" posture
* the data file can be load by ```pickle```, e.g. ```pickle.load(open('Back', 'rb'))```, the data is a list of feature data
* the features (e.g. each row of the data) are ['LHipYawPitch', 'LHipRoll', 'LHipPitch', 'LKneePitch', 'RHipYawPitch', 'RHipRoll', 'RHipPitch', 'RKneePitch', 'AngleX', 'AngleY'], where 'AngleX' and 'AngleY' are body angle (e.g. ```Perception.imu```) and others are joint angles.

## 2. Data preprocessing

````python
%matplotlib inline
import pickle
from os import listdir, path
import numpy as np
from sklearn import svm, metrics
from sklearn.model_selection import train_test_split

ROBOT_POSE_DATA_DIR = 'robot_pose_data'
````

````python
classes = listdir(ROBOT_POSE_DATA_DIR)
print(classes)
````

````python
def load_pose_data(i):
    '''load pose data from file'''
    data = []
    target = []
    # YOUR CODE HERE
    
    filename = path.join(ROBOT_POSE_DATA_DIR, classes[i])
    data = pickle.load(open(filename, 'rb'))
    target = [i] * len(data)
    return data, target
````

````python
# load all the data
all_data = []
all_target = []
# YOUR CODE HERE
for i, joint in enumerate(classes):
    data, target = load_pose_data(i)
    all_data.extend(data)
    all_target.extend(target)
    print(joint, len(data), len(target))
    
print ('total number of data', len(all_data))
````

````python
# shuffule data
permutation = np.random.permutation(len(all_data))
n_training_data = int(len(all_data) * 0.7)
training_data = permutation[:n_training_data]
````

## 3. Learn on training data

In scikit-learn, an estimator for classification is a Python object that implements the methods fit(X, y) and predict(T). An example of an estimator is the class sklearn.svm.SVC that implements support vector classification.

````python
clf = svm.SVC(gamma=0.001, C=100.)
````

### learning

````python
# YOUR CODE HERE
X_train, X_test, y_train, y_test = train_test_split(all_data, all_target, test_size=0.25)
print(len(X_train))
print(len(X_test))
clf.fit(X_train, y_train)
````

### predicting

````python
clf.predict(all_data[-1:]), all_target[:-1:]
````

````python
def evaluate(expected, predicted):
    print("Classification report:\n%s\n" % metrics.classification_report(expected, predicted))

    print("Confusion matrix:\n%s" % metrics.confusion_matrix(expected, predicted))
````

````python
expected = []
predicted = []
# YOUR CODE HERE
expected = y_train
predicted = clf.predict(X_train)
evaluate(expected, predicted)
````

## 4. Evaluate on the test data

````python
expected = []
predicted = []
# YOUR CODE HERE
expected = y_test
predicted = clf.predict(X_test)
evaluate(expected, predicted)
````

## 5. Deploy to the real system

We can simple use `pickle` module to serialize the trained classifier.

````python
ROBOT_POSE_CLF = 'robot_pose.pkl'
pickle.dump(clf, open(ROBOT_POSE_CLF, 'wb'))
````

Then, in the application we can load the trained classifier again.

````python
with open(ROBOT_POSE_CLF, 'rb') as f:
    clf2 = pickle.load(f)
clf2.predict(all_data[:-1]), all_target[:-1]
````
